**Broadcasting** is  automatically applying a number to every element of matrices. **For Example** Lets create a 3*3 matrix and consider our number is 2

## Part B

### System 1
Solve the following two system of linear equations using matrices (find the values of x1 and x2) and 
      determine the number of solutions using the functions.

         1.2x1+3x2-4x3=6,x1-4x2=8
	
         2. 3y1-4y2+5y3=10,-y1+2y2-4y3=8

#### Solution
The rank of the coefficient matrix and the augmented matrix is used to determine the number of solutions.We need to convert equations into matrices( A1X = B1). Which gives us A1 =
[[ 2,  3, -4],
 [ 1, -4,  0]] and
  B1 =
[6, 8] and
X = [x1, x2, x3]

```python
A1 = np.array([
    [2, 3, -4],
    [1, -4, 0]
])

B1 = np.array([6, 8])

print("Coefficient Matrix A:")
print(A1)

print("\nConstant Matrix B:")
print(B1)
```

To find the solution

```python

# STEP 1: Define the Coefficient Matrix (A) and Vector (B) 
# Setting x3 = 0 to get 2 equations for 2 variables (x1, x2)
A1 = np.array([[2, 3],
               [1, -4]])

B1 = np.array([6, 8])

#  STEP 2: Check Number of Solutions using Rank 
# Build full matrix including x3 and constant term
A1_full = np.array([[2, 3, -4],
                    [1, -4, 0]])

A1_aug = np.array([[2, 3, -4, 6],
                   [1, -4, 0, 8]])

rank_A1 = np.linalg.matrix_rank(A1_full)
rank_aug1 = np.linalg.matrix_rank(A1_aug)

print("Rank of A:", rank_A1)
print("Rank of Augmented:", rank_aug1)

if rank_A1 == rank_aug1:
    print("Number of solutions: Infinitely many solutions (since rank < 3 variables)")

# --- STEP 3: Solve for x1 and x2 ---
# Solve A1 * X = B1 assuming x3 = 0
X1 = np.linalg.solve(A1, B1)

print("Value of x1:", X1[0])
print("Value of x2:", X1[1])
```

### Result for System 1

Number of solutions: Infinitely many solutions   
Values (when $x_3 = 0$): $x_1 \approx 4.36$, $x_2 \approx -0.91$

### System 2

The system of equations is:

3y1 - 4y2 + 5y3 = 10

-y1 + 2y2 - 4y3 = 8

Using similar coefficient technique as used in system 1

```python
A2 = np.array([
    [3, -4, 5],
    [-1, 2, -4]
])

B2 = np.array([10, 8])

print("Coefficient Matrix A:")
print(A2)

print("\nConstant Matrix B:")
print(B2)
```

To find the solution

```python
# STEP 1: Define the Coefficient Matrix (A) and Vector (B) 
# Setting y3 = 0 to get 2 equations for 2 variables (y1, y2)
A2 = np.array([[ 3, -4],
               [-1,  2]])

B2 = np.array([10, 8])

# --- STEP 2: Check Number of Solutions using Rank ---
A2_full = np.array([[ 3, -4, 5],
                    [-1,  2, -4]])

A2_aug = np.array([[ 3, -4, 5, 10],
                   [-1,  2, -4, 8]])

rank_A2 = np.linalg.matrix_rank(A2_full)
rank_aug2 = np.linalg.matrix_rank(A2_aug)

print("Rank of A:", rank_A2)
print("Rank of Augmented:", rank_aug2)

if rank_A2 == rank_aug2:
    print("Number of solutions: Infinitely many solutions (since rank < 3 variables)")

# --- STEP 3: Solve for y1 and y2 ---
# Solve A2 * Y = B2 assuming y3 = 0
Y2 = np.linalg.solve(A2, B2)

print("Value of y1:", Y2[0])
print("Value of y2:", Y2[1])
```

### Result for System 2

Number of solutions: Infinitely many solutions   Values (when $y_3 = 0$): $y_1 = 26$, $y_2 = 17$
